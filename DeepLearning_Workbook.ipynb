# 📘 Module 1: PyTorch 기초 문법 & 텐서 연산

PyTorch의 핵심 데이터 구조는 **Tensor(텐서)**입니다. NumPy의 `ndarray`와 매우 유사하지만, **GPU/TPU 가속**과 **자동 미분(Autograd)**을 지원합니다.

### 1. 텐서 속성 (Shape, Dtype, Device)
- `shape` 또는 `size()`: 텐서의 형태 (각 차원의 크기)
- `dtype`: 텐서의 데이터 타입 (`float32`, `int64`, `bool` 등)
- `device`: 텐서가 위치한 장치 (`cpu`, `cuda`, `mps`)

### 2. 행렬 연산 주의점
- 원소별 곱셈 (Element-wise Multiplication): `A * B` 또는 `torch.mul(A, B)`
- 행렬 곱 (Matrix Multiplication): `A @ B` 또는 `torch.matmul(A, B)`
- **행렬 곱 조건**: $A \in \mathbb{R}^{M 	imes N}$ 이고 $B \in \mathbb{R}^{N 	imes K}$ 일 때 결과는 $(M 	imes K)$ 모양이어야 함.

# 📐 Module 2: 차원 관리 Masterclass (Dimension Management)

딥러닝 모델 구현 중 발생하는 에러의 **80% 이상은 차원(Shape) 불일치**에서 비롯됩니다!
차원 조작 함수들의 특성과 올바른 사용법을 완벽히 마스터해야 합니다.

### 핵심 개념 정리
| 함수 | 역할 | 비고 |
|---|---|---|
| `view(*shape)` | 차원 변경 (메모리 연속일 때만) | 빠른 속도, 메모리 공유 |
| `reshape(*shape)` | 차원 변경 (연속성 상관없음) | 필요 시 메모리 복사 발생 |
| `unsqueeze(dim)` | 특정 위치에 차원(크기 1) 추가 | `(A, B)` ➔ `(1, A, B)` (dim=0) |
| `squeeze(dim)` | 크기가 1인 차원 제거 | `(1, A, B)` ➔ `(A, B)` |
| `transpose(dim0, dim1)` | 두 차원의 위치 교환 | 2개 차원 교환 |
| `permute(*dims)` | 모든 차원의 순서 재배치 | 이미지 `(B, H, W, C)` ➔ `(B, C, H, W)` |
| `torch.cat(tensors, dim)` | 기존 차원을 기준으로 결합 | 차원 수 유지 |
| `torch.stack(tensors, dim)` | 새로운 차원을 생성하며 결합 | 차원 수 +1 증가 |

# ⚙️ Module 3: 자동 미분(Autograd) & 신경망 모듈 (nn.Module)

PyTorch의 **Autograd Engine**은 테이프 기반 역전파(Backpropagation) 시스템을 통해 모델 파라미터의 Gradient(경사도)를 자동으로 계산합니다.

### Autograd 사용 법
1. `requires_grad=True` 설정: 미분 대상 텐서 지정
2. 손실(Loss) 스칼라 값 계산
3. `loss.backward()` 호출: 역전파 실행 및 `.grad` 속성에 경사도 누적
4. `optimizer.zero_grad()` 또는 `tensor.grad.zero_()`: 기존 누적된 경사도 초기화

### PyTorch `nn.Module` 구조 기본 틀
```python
class CustomModel(nn.Module):
    def __init__(self):
        super().__init__()
        # 레이어 정의
        self.fc1 = nn.Linear(in_features, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, out_features)

    def forward(self, x):
        # 순전파 연산 정의
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x
```

# 🎯 Module 4: 학습 전략 및 실전 기법 (Training Strategies)

실전 딥러닝 모델 학습 시 필수적으로 고려해야 하는 요소들입니다.

### 1. 손실 함수 (Loss Functions)
- **`nn.MSELoss`**: 회귀(Regression) 문제에 사용. 입력과 타겟의 shape이 동일해야 함.
- **`nn.CrossEntropyLoss`**: 다중 클래스 분류(Multi-class Classification) 문제에 사용.
  - **주의!**: PyTorch의 `CrossEntropyLoss`는 내부적으로 **Softmax 연산을 포함**합니다! 따라서 모델의 마지막 레이어에 Softmax를 적용하지 않은 **Logits** 상태의 텐서를 입력으로 넣어야 합니다.
  - **입력 Logits Shape**: `(Batch_Size, Num_Classes)`
  - **타겟 Target Shape**: `(Batch_Size,)`이며 `dtype`은 **`torch.long` (정수 클래스 인덱스)** 이어야 합니다.

### 2. 정규화 & 드롭아웃 (Regularization & Dropout)
- **`nn.BatchNorm1d` / `nn.BatchNorm2d`**: 배치 정규화. 학습 안정성 향상 및 빠른 수렴 지원.
- **`nn.Dropout(p=0.5)`**: 학습 과정에서 랜덤하게 노드를 끄는 기법 (오버피팅 방지).
- **`model.train()` vs `model.eval()`**:
  - `BatchNorm`과 `Dropout`은 학습 시(`train()`)와 평가/추론 시(`eval()`) 다르게 동작하므로, Validation/Test 루프 진입 전 반드시 **`model.eval()`**을 호출하고 **`with torch.no_grad():`** 블록 안에서 실행해야 합니다!
